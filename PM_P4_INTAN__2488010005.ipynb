# Langkah 1


```python
import pandas as pd
import numpy as np

# Membuat dataframe sampel
data = {
    'usia': [25, 30, np.nan, 45, 22, 38, np.nan, 29],
    'pendapatan': [5000000, 7500000, 6000000, np.nan, 4000000, 12000000, 8000000, np.nan],
    'pendidikan': ['SMA', 'S1', 'S2', 'S1', 'SMA', 'S2', 'S1', 'SMA'],
    'kota': ['Jakarta', 'Bandung', 'Surabaya', 'Jakarta', 'Bandung', 'Jakarta', 'Surabaya', 'Bandung'],
    'membeli': ['Ya', 'Tidak', 'Ya', 'Ya', 'Tidak', 'Ya', 'Tidak', 'Ya']
}

df = pd.DataFrame(data)

df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
```

## langkah 2

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

## langkah 3

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
```

## langkah 4

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)

```

## langkah 5

```python
from sklearn.preprocessing import StandardScaler

num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()

X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])    # cegah data leakage
```

## langkah 6

```python
print("Nilai kosong:")
print(X_train.isnull().sum())

print("\nTipe data:")
print(X_train.dtypes)

print("\nStatistik:")
print(X_train.describe())
```
